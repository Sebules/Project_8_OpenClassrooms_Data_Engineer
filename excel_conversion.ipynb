```python
INPUT_FILE = "https://s3.eu-west-1.amazonaws.com/course.oc-static.com/projects/922_Data+Engineer/922_P8/Weather+Underground+-+Ichtegem%2C+BE.xlsx"
OUTPUT_FILE = "Weather_Underground_Ichtegem_BE.csv"
```

```python
sheets = pd.read_excel(INPUT_FILE,sheet_name=None)
```

```python
sheets.items()
```

```python
frames = []
for tab_name, df in sheets.items():
    df["tab_name"] = tab_name          
    frames.append(df)
```

```python
result = pd.concat(frames, ignore_index=True)
```

```python
result.head()
```

```python
result["Date"] = pd.to_datetime(result["tab_name"], format="%d%m%y")
```

```python
result
```

```python
result.info()
```

```python
columns = ['Time', 'Temperature', 'Humidity', 'Wind', 'Speed', 'Gust',
           'Pressure', 'Precip. Rate.', 'Precip. Accum.','UV', 'Solar']

result = result.dropna(how='all', axis=0, subset=columns,ignore_index=True)
```

```python
result.info()
```

```python
result.to_csv('donnees/'+ OUTPUT_FILE, encoding='UTF-8', index=False)
```
